```python
total_sales_California = sales_data[sales_data["State"] == "California"]["Sales"].sum()
```

```python
total_sales_California
```

```python
total_sales_by_state = sales_data.groupby("State")["Sales"].sum().reset_index()

total_sales_by_state
```

```python
total_sales_by_state.sort_values(by="Sales", ascending=False, inplace=True)

total_sales_by_state
```

```python
avg_sales_by_state = sales_data.groupby("State")["Sales"].mean().reset_index()

avg_sales_by_state.sort_values(by="Sales", ascending=False, inplace=True)

avg_sales_by_state
```

```python
avg_total_sales = sales_data.groupby("State")["Sales"].agg([np.sum, np.mean, np.max]).reset_index()

avg_total_sales
```
